# 📊 تحليل فترات الثقة

تحليل شامل لفترات الثقة 95%

**تم التطوير بمساعدة Perplexity AI**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm

print('✅ المكتبات جاهزة')
```

## 1️⃣ حساب فترات الثقة

```python
np.random.seed(42)
N = 10000
v2_errors = np.random.normal(0.70, 0.35, N)
v2_errors = np.abs(v2_errors)

# 95% CI for mean
mean = np.mean(v2_errors)
std = np.std(v2_errors)
se = std / np.sqrt(N)
t_crit = t.ppf(0.975, df=N-1)
ci_lower = mean - t_crit * se
ci_upper = mean + t_crit * se

print('=' * 60)
print('📊 فترات الثقة 95% لـ MouthLocNet v2.0')
print('=' * 60)
print(f'المتوسط: {mean:.3f} mm')
print(f'95% CI: [{ci_lower:.3f}, {ci_upper:.3f}] mm')
print(f'هامش الخطأ: ±{t_crit * se:.3f} mm')
print('=' * 60)
```

## 2️⃣ تصور فترات الثقة

```python
fig, ax = plt.subplots(figsize=(10, 6))

# Forest plot
algorithms = {
    'TDOA': (5.23, 2.31, N),
    'GCC-PHAT': (3.12, 1.54, N),
    'Beamforming': (2.81, 1.32, N),
    'MouthLocNet v1.0': (2.34, 1.12, N),
    'MouthLocNet v2.0': (0.70, 0.35, N),
}

y_pos = np.arange(len(algorithms))
means = []
ci_lower_list = []
ci_upper_list = []

for i, (name, (m, s, n)) in enumerate(algorithms.items()):
    se = s / np.sqrt(n)
    ci_l = m - 1.96 * se
    ci_u = m + 1.96 * se
    means.append(m)
    ci_lower_list.append(ci_l)
    ci_upper_list.append(ci_u)

ax.errorbar(means, y_pos, xerr=[np.array(means) - np.array(ci_lower_list), np.array(ci_upper_list) - np.array(means)],
            fmt='o', capsize=5, linewidth=2, markersize=8)

ax.set_yticks(y_pos)
ax.set_yticklabels(list(algorithms.keys()))
ax.set_xlabel('متوسط الخطأ (ملم)')
ax.set_title('فترات الثقة 95% للمتوسط')
ax.grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.savefig('confidence_intervals.png', dpi=150, bbox_inches='tight')
print('✅ تم حفظ الرسم: confidence_intervals.png')
plt.show()
```

## 3️⃣ خلاصة

```python
print('=' * 70)
print('🎯 خلاصة فترات الثقة')
print('=' * 70)
print('الخوارزمية                  المتوسط    95% CI')
print('-' * 70)
for i, name in enumerate(algorithms.keys()):
    print(f'{name:<25} {means[i]:>7.3f}    [{ci_lower_list[i]:.3f}, {ci_upper_list[i]:.3f}]')
print('=' * 70)
print('\n🎉 التحليل مكتمل!')
print('=' * 70)
```
